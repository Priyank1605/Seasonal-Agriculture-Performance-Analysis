```python
#IMPORT ALL REQ LIBRARIES
#In this step, we import the Python libraries required for data analysis and visualization.
#Pandas is used for handling data, NumPy for numerical operations, and Matplotlib and Seaborn for creating graphs.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
```

```python
#LOAD DATASET
#Here, we load the agricultural dataset using Pandas.
#The dataset contains information about farms, crops, seasons, environmental conditions, resource usage, production and economic performance.
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)
```

```python
#Check Missing Values
#In this step, we check whether any values are missing from the dataset.
#Missing values can affect the accuracy of analysis, so they need to be identified before further processing.
df.isnull().sum()
```

```python
#Check Duplicate Records
#Duplicate records can affect analysis by counting the same observation more than once. Therefore, we check the dataset for duplicate rows.
df.duplicated().sum()
```

```python
#Remove Duplicate Records
#If duplicate records are present, they can be removed to maintain the quality and accuracy of the analysis.
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
#Check Categorical Variables
#This step shows how many agricultural records belong to each season.
#It helps us understand whether the dataset contains balanced information across seasons.
print(df["Season"].value_counts())
```

```python
#data _cleaning
df = df.drop_duplicates()

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
#categorical column
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#Distribution of yield
plt.figure(figsize=(2,2))
sns.histplot(df["Yield_Tonnes_Ha"], kde=True)
plt.title("Distribution of Crop Yield")
plt.xlabel("Yield (Tonnes/Ha)")
plt.ylabel("Frequency")
plt.show()
```

```python
#distribution of agricultural profit
plt.figure(figsize=(2,2))
sns.histplot(df["Profit_INR"], kde=True)
plt.title("Distribution of Agricultural Profit")
plt.xlabel("Profit (INR)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Seasonal Analysis, This is the most important section of your project.
#The project brief specifically asks you to investigate how agricultural performance varies across seasons.
season_count = df["Season"].value_counts()

season_count.plot(kind="bar", figsize=(2,2))

plt.title("Number of Farms by Season")
plt.xlabel("Season")
plt.ylabel("Number of Farms")
plt.xticks(rotation=0)
plt.show()
```

```python
#avg yeild by season
season_yield = df.groupby("Season")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)
print(season_yield)
season_yield.plot(kind="bar", figsize=(3,3))
plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=0)
plt.show()
```

```python
#Production by Season
season_production = df.groupby("Season")["Production_Tonnes"].sum()
print(season_production)
season_production.plot(kind="bar", figsize=(2,2))
plt.title("Total Agricultural Production by Season")
plt.xlabel("Season")
plt.ylabel("Production (Tonnes)")
plt.xticks(rotation=0)
plt.show()
```

```python
#reveneve by season
season_revenue = df.groupby("Season")["Revenue_INR"].sum()

print(season_revenue)
```

```python
#profit by season
season_profit = df.groupby("Season")["Profit_INR"].sum()
print(season_profit)
season_profit.plot(kind="bar", figsize=(3,3))
plt.title("Total Profit by Season")
plt.xlabel("Season")
plt.ylabel("Profit (INR)")
plt.xticks(rotation=0)
plt.show()
```

```python
#Average Profit by Season
avg_profit = df.groupby("Season")["Profit_INR"].mean()

print(avg_profit)
```

```python
#Cost by Season
season_cost = df.groupby("Season")["Total_Cost_INR"].mean()
print(season_cost)
```

```python
#Water Usage by Season
water_season = df.groupby("Season")["Water_Used_m3"].mean()
print(water_season)
```

```python
#Water Efficiency by Season
water_efficiency = df.groupby("Season")[
    "Water_Efficiency_t_per_1000m3"
].mean()
print(water_efficiency)
```

```python
#Disease/Pest Risk by Season
risk_season = df.groupby("Season")["Disease_Pest_Risk_pct"].mean()
print(risk_season)
risk_season.plot(kind="bar", figsize=(3,3))
plt.title("Average Disease and Pest Risk by Season")
plt.xlabel("Season")
plt.ylabel("Risk (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
#Environmental Analysis Rainfall, Temperature, Humidity, Sunlight, Soil moisture, Soil pH
rainfall = df.groupby("Season")["Rainfall_mm"].mean()
rainfall.plot(kind="bar", figsize=(3,3))
plt.title("Average Rainfall by Season")
plt.xlabel("Season")
plt.ylabel("Rainfall (mm)")
plt.show()
```

```python
#Temperature by Season
temperature = df.groupby("Season")["Avg_Temperature_C"].mean()
temperature.plot(kind="bar", figsize=(3,3))
plt.title("Average Temperature by Season")
plt.xlabel("Season")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
#Soil Moisture by Season
soil_moisture = df.groupby("Season")["Soil_Moisture_pct"].mean()
print(soil_moisture)
```

```python
#Crop-wise Analysis, Crop distribution
crop_count = df["Crop"].value_counts()
crop_count.plot(kind="bar", figsize=(3,3))
plt.title("Crop Distribution")
plt.xlabel("Crop")
plt.ylabel("Number of Farms")
plt.xticks(rotation=45)
plt.show()
```

```python
#Average Yield by Crop
crop_yield = df.groupby("Crop")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)
print(crop_yield)
```

```python
#Crop × Season Analysis
#This analysis compares crop yield across different seasons. It helps identify which crops perform better under specific seasonal conditions.
crop_season_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="Crop",
    columns="Season",
    aggfunc="mean")
print(crop_season_yield)
plt.figure(figsize=(3,3))
sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt=".2f")
plt.title("Average Crop Yield Across Seasons")
plt.xlabel("Season")
plt.ylabel("Crop")
plt.show()
```

```python
#Irrigation Analysis
#Different irrigation methods may influence crop productivity. Here, we compare the average yield associated with each irrigation method.
irrigation_yield = df.groupby(
    "Irrigation_Method"
)["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)
print(irrigation_yield)
irrigation_yield.plot(kind="bar", figsize=(3,3))
plt.title("Average Yield by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.show()
```

```python
#Irrigation and Water Usage
#This analysis compares average water consumption for different irrigation methods. It helps us understand resource usage.
irrigation_water = df.groupby(
    "Irrigation_Method"
)["Water_Used_m3"].mean().sort_values()
print(irrigation_water)
```

```python
#Fertilizer Analysis
#Fertilizer is an important agricultural input. This step compares average fertilizer usage across seasons.
fertilizer_season = df.groupby("Season")["Fertilizer_kg_ha"].mean()
print(fertilizer_season)
```

```python
#Pesticide Analysis
pesticide_season = df.groupby("Season")["Pesticide_Litre_ha"].mean()
print(pesticide_season)
```

```python
#Seed Quality Analysis
#Seed quality can influence crop performance. We calculate the average seed quality score for each season.
seed_quality = df.groupby("Season")["Seed_Quality_Score"].mean()
print(seed_quality)
```

```python
#Economic Analysis: Market price, Total cost, Revenue, Profit
#These should be a major part of your project.
economic = df.groupby("Season")[
    ["Revenue_INR", "Total_Cost_INR", "Profit_INR"]
].mean()
print(economic)
```

```python
#Profit Margin
#Create a new feature:
df["Profit_Margin_pct"] = (
    df["Profit_INR"] / df["Revenue_INR"]
) * 100
#Check:
df["Profit_Margin_pct"].describe()
#Season-wise:
profit_margin = df.groupby("Season")["Profit_Margin_pct"].mean()
print(profit_margin)
```

```python
#Market Price by Season
market_price = df.groupby("Season")[
    "Market_Price_INR_Tonne"
].mean()
print(market_price)
```

```python
#Best Season Overall

season_summary = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "sum",
    "Revenue_INR": "mean",
    "Total_Cost_INR": "mean",
    "Profit_INR": "mean",
    "Water_Used_m3": "mean",
    "Water_Efficiency_t_per_1000m3": "mean",
    "Disease_Pest_Risk_pct": "mean"
})

season_summary
```

```python
#Create Overall Performance Score
from sklearn.preprocessing import MinMaxScaler

score_columns = [
    "Yield_Tonnes_Ha",
    "Profit_INR",
    "Water_Efficiency_t_per_1000m3"
]

season_score = df.groupby("Season")[score_columns].mean()

scaler = MinMaxScaler()

scaled = scaler.fit_transform(season_score)

season_score["Performance_Score"] = scaled.mean(axis=1)

season_score = season_score.sort_values(
    "Performance_Score",
    ascending=False
)

season_score
```

```python
#State-wise Analysis
state_summary = df.groupby("State").agg({
    "Yield_Tonnes_Ha": "mean",
    "Profit_INR": "mean",
    "Revenue_INR": "mean",
    "Water_Efficiency_t_per_1000m3": "mean"
}).sort_values(
    "Profit_INR",
    ascending=False
)

state_summary
```

```python
#District-wise Analysis
district_summary = df.groupby("District").agg({
    "Yield_Tonnes_Ha": "mean",
    "Profit_INR": "mean"
}).sort_values(
    "Profit_INR",
    ascending=False
)

district_summary.head(10)
```

```python
#Season × State Analysis
state_season = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="State",
    columns="Season",
    aggfunc="mean"
)

plt.figure(figsize=(12,8))

sns.heatmap(
    state_season,
    annot=True,
    fmt=".2f"
)

plt.title("Average Yield by State and Season")
plt.show()
```
